### Phase 2: Interactive Data Cleaning

**User Guidance (Column Definitions):**

You have provided the following definitions for the key columns:
*   `HIERARCHY_TYPE`: Defines if the item is a `Recipe`, `Sub-recipe`, etc.
*   `PRODUCTO`: The name of the recipe or sub-recipe.
*   `INGREDIENTES`: The generic name of an ingredient.
*   `SKU_MATCH`: The specific, purchasable SKU that maps to the generic ingredient. This is the key we will use to find the price.
*   `CANTIDAD`: The quantity of the ingredient needed.
*   `MEDIDA`: The unit of measure for the quantity.

**Goal:** Clean and standardize the `Recetas` and `Inventario` sheets based on these rules.


**Action A: Select and Rename Columns in `Recetas` Sheet**

First, I will select only the columns we need and rename them to be more code-friendly.


```python
# Define the columns we want to keep and their new names
columns_to_keep = {
    'HIERARCHY_TYPE': 'hierarchy_type',
    'PRODUCTO': 'recipe_name',
    'INGREDIENTES': 'ingredient_name',
    'CANTIDAD': 'quantity',
    'MEDIDA': 'unit',
    'SKU_MATCH': 'sku_match'
}

# Select and rename the columns
try:
    cleaned_recetas_df = recetas_df[list(columns_to_keep.keys())].copy()
    cleaned_recetas_df.rename(columns=columns_to_keep, inplace=True)

    print("✅ Successfully selected and renamed columns.")
    print("📋 Preview of the cleaned 'Recetas' DataFrame:")
    display(cleaned_recetas_df.head())
    
except KeyError as e:
    print(f"❌ ERROR: A specified column was not found in the DataFrame. Please check the column names. Details: {e}")
except Exception as e:
    print(f"❌ ERROR: An unexpected error occurred: {e}")
```

**Action B: Clean the `quantity` Column**

Next, I will convert the `quantity` column to a numeric data type. I will then find and display any rows where the quantity could not be converted, so you can decide how we should handle them.


```python
# Convert 'quantity' to numeric, coercing errors to NaN
cleaned_recetas_df['quantity'] = pd.to_numeric(cleaned_recetas_df['quantity'], errors='coerce')

# Find rows where quantity could not be converted
invalid_quantity_rows = cleaned_recetas_df[cleaned_recetas_df['quantity'].isna()]

if invalid_quantity_rows.empty:
    print("✅ All quantity values were successfully converted to a numeric type.")
else:
    print(f"⚠️ Found {len(invalid_quantity_rows)} rows where 'quantity' could not be converted to a number.")
    print("   Please review these rows:")
    display(invalid_quantity_rows)

print("\\n📋 Technical info for the DataFrame after conversion:")
cleaned_recetas_df.info()
```

**Insight into Missing Quantities**

To help decide how to handle the rows with missing quantities, I will now analyze the rest of the dataset. For each ingredient with a missing quantity, I will calculate the average quantity used for that same ingredient in other recipes. This will provide context for our decision.


```python
# Calculate the average quantity for each ingredient across the entire dataset
avg_quantities = cleaned_recetas_df.groupby('ingredient_name')['quantity'].mean().to_dict()

# Map these average quantities to the rows with invalid quantities
invalid_quantity_rows['avg_quantity_elsewhere'] = invalid_quantity_rows['ingredient_name'].map(avg_quantities)

print("📋 Context for ingredients with missing quantities:")
display(invalid_quantity_rows[['recipe_name', 'ingredient_name', 'avg_quantity_elsewhere']])
```

**User Guidance: Handling Missing Quantities**

You have directed me to handle the 93 rows with missing quantities by filling them with `0`. This is a safe approach that keeps the ingredient record in the recipe without affecting the cost calculation. The long-term plan is to build a dedicated "Recipe Builder" tool to properly manage and fill in these gaps.

I will now apply this rule.


```python
# Fill NaN values in the 'quantity' column with 0
cleaned_recetas_df['quantity'].fillna(0, inplace=True)

# Verify that there are no more NaN values in the 'quantity' column
nan_in_quantity = cleaned_recetas_df['quantity'].isna().sum()

if nan_in_quantity == 0:
    print("✅ Successfully filled all missing quantity values with 0.")
    print("   The 'quantity' column is now clean.")
else:
    print(f"⚠️ Verification failed. There are still {nan_in_quantity} missing values in the 'quantity' column.")

print("\\n📋 Technical info for the DataFrame after filling missing values:")
cleaned_recetas_df.info()
```

**Action C: Standardize Units of Measure**
